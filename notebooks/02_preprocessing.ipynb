```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv("../data/ieee-fraud-detection/train_transaction.csv")
identity_df = pd.read_csv("../data/ieee-fraud-detection/train_identity.csv")
df_merged = df.merge(identity_df, on="TransactionID", how="left")
print(df_merged.shape)
```

```python
#drop column who has more than 50% null values
df_merged = df_merged.dropna(thresh=len(df_merged) * 0.5, axis=1)
df_merged.shape
```

```python
nums_cols = df_merged.select_dtypes(include=[np.number]).columns
print(nums_cols)
```

```python
df_merged[nums_cols] = df_merged[nums_cols].fillna(df_merged[nums_cols].median())
```

```python
df_merged.isnull().sum()
```

```python
cat_cols = df_merged.select_dtypes(include=['object']).columns
print(cat_cols)
```

```python
cat_cols.fillna(df[cat_cols].mode().iloc[0])
```

```python
for col in cat_cols:
    le = LabelEncoder()
    df_merged[col] = le.fit_transform(df_merged[col])
```

```python
df_merged.to_csv("../data/processed/cleaned_data.csv", index=False)
```
